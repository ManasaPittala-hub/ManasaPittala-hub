```python
completed_tasks = []
incomplete_tasks = []

tasks = ["Study Python", "Complete Assignment", "Exercise"]

for task in tasks:
    status = input("Did you complete " + task + "? (yes/no): ")

    if status.lower() == "yes":
        completed_tasks.append(task)
    else:
        incomplete_tasks.append(task)

print("Completed Tasks:", completed_tasks)
print("Incomplete Tasks:", incomplete_tasks)
```
